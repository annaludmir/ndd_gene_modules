### fisher result enrichments in the sig tables

```python
#function
from scipy.stats import hypergeom
from scipy.stats import fisher_exact

def fisher_test_hsg(hsg_table,sig_table,n_genes=30251,n_hsg=443):
    hsg_intersection=list(set(sig_table["gene"]) & set(hsg_table["gene"]))
    table=np.array([[len(hsg_intersection),n_hsg-len(hsg_intersection)],
                    [len(sig_table)-len(hsg_intersection),n_genes-len(sig_table)-len(hsg_intersection)]])
    res = fisher_exact(table, alternative='two-sided')
    return (hsg_intersection , res.pvalue)
```

```python
fisher_test_hsg(hsg,sig_ges_scores_v3_rg)
```

```python
def fisher_test_har(har_table,sig_table,n_genes=30251,n_har=701):
    har_intersection=list(set(sig_table["gene"]) & set(har_table["HAR-associated genes.1"]))
    table=np.array([[len(har_intersection),n_har-len(har_intersection)],
                    [len(sig_table)-len(har_intersection),n_genes-len(sig_table)-len(har_intersection)]])
    res = fisher_exact(table, alternative='two-sided')
    return (har_intersection , res.pvalue)
```

```python
fisher_test_har(har_genes,sig_ges_scores_v3_rg) #almost significant but not
```
