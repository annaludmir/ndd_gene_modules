```python
microcephaly=pd.read_csv("/scratch200/reutj/data/severe_microcephaly_genes.csv")
```

```python
microcephaly
```

```python
res=pd.DataFrame(mg.querymany(microcephaly["Gene Symbol"], scopes='symbol', fields='ensembl.gene', species='human'))
res
```

```python
#function to gather the hits for each gene together 
dup_dict={}
for i in range(len(res)):
    gene=res['query'][i]
    dup_dict[gene]=[]
for i in range(len(res)):
    gene=res['query'][i]
    ens=res.ensembl[i]
    if type(ens)==dict:
        dup_dict[gene].append(ens['gene'])
    elif type(ens)==list:
        dup_dict[gene]=ens

dup_dict
```

```python
microcephaly_df1=pd.DataFrame.from_dict(dup_dict,orient='index').iloc[:268,:1]#drop last nan line
microcephaly_df1=microcephaly_df1.rename(columns={0:'ens'})
microcephaly_df1
```

```python
#change everything
microcephaly_df1.loc[microcephaly_df1.index=="AARS",'ens']= "ENSG00000090861"
microcephaly_df1.loc[microcephaly_df1.index=="CENPJ",'ens']="ENSG00000151849"
microcephaly_df1.loc[microcephaly_df1.index=="HIST1H4C",'ens']="ENSG00000197061"
microcephaly_df1.loc[microcephaly_df1.index=="IARS",'ens']="ENSG00000196305"
microcephaly_df1.loc[microcephaly_df1.index=="KIF1BP",'ens']="ENSG00000198954"
microcephaly_df1.loc[microcephaly_df1.index=="NARS",'ens']="ENSG00000134440"
microcephaly_df1.loc[microcephaly_df1.index=="SARS",'ens']="ENSG00000031698"
microcephaly_df1.loc[microcephaly_df1.index=='SPATA5L1','ens']="ENSG00000171763"
microcephaly_df1.loc[microcephaly_df1.index=='C7orf43','ens']="ENSG00000146826"
microcephaly_df1.loc[microcephaly_df1.index=='DYNC1I2','ens']="ENSG00000077380"
microcephaly_df1.loc[microcephaly_df1.index=='EIF5A','ens']="ENSG00000132507"
microcephaly_df1.loc[microcephaly_df1.index=='HHAT','ens']="ENSG00000054392"
microcephaly_df1.loc[microcephaly_df1.index=='NDE1','ens']="ENSG00000072864"
microcephaly_df1.loc[microcephaly_df1.index=='OSGEP','ens']="ENSG00000092094"
microcephaly_df1.loc[microcephaly_df1.index=='TOP3A','ens']="ENSG00000177302"
microcephaly_df1.loc[microcephaly_df1.index=='TRAPPC12','ens']="ENSG00000171853"
microcephaly_df1.loc[microcephaly_df1.index=='TTC5','ens']="ENSG00000136319"
microcephaly_df1.loc[microcephaly_df1.index=='WDR73','ens']="ENSG00000177082"
microcephaly_df1.loc[microcephaly_df1.index=='AKT3','ens']="ENSG00000117020"
microcephaly_df1.loc[microcephaly_df1.index=='TPRKB','ens']="ENSG00000144034"
microcephaly_df1.loc[microcephaly_df1.index=='TUBGCP3','ens']="ENSG00000126216"
microcephaly_df1.loc[microcephaly_df1.index=='ZNHIT3','ens']="ENSG00000273611"
microcephaly_df1.loc[microcephaly_df1.index=='RING1','ens']='ENSG00000231115'
```

```python
microcephaly_df2=pd.DataFrame({'gene':microcephaly_df1.index,'ens':microcephaly_df1.ens}).reset_index()
microcephaly_df2=microcephaly_df2.drop(columns='index')
microcephaly_df2
```

```python
#just change names that are difenr than gencards annotations
microcephaly_df2.loc[microcephaly_df2.gene=="AARS",'gene']= "AARS1"
microcephaly_df2.loc[microcephaly_df2.gene=="CENPJ",'gene']= "CPAP"
microcephaly_df2.loc[microcephaly_df2.gene=="IARS",'gene']= "IARS1"
microcephaly_df2.loc[microcephaly_df2.gene=="NARS",'gene']= "NARS1"
microcephaly_df2.loc[microcephaly_df2.gene=="SARS",'gene']= "SARS1"
microcephaly_df2.loc[microcephaly_df2.gene=="SPATA5L1",'gene']= "AFG2B"
microcephaly_df2.loc[microcephaly_df2.gene=="C7orf43",'gene']= "TRRAPC14"
```

```python
#SAVE TO CSV
microcephaly_df2.to_csv("/scratch200/reutj/data/hsg_gene_lists/microcephaly_genes.csv")
```

```python
#to gmt
output_file="/scratch200/reutj/data/hsg_gene_lists/microcephaly_genes.gmt"
with open(output_file, 'w') as f:
    # Create a line in GMT format for one gene set
    line = f"microcephaly_genes\tDescription\t{'\t'.join(microcephaly_df2.gene.tolist())}\n"
    f.write(line)
```
