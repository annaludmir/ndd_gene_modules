## unify data with harmony

```python
import scanpy as sc
import pandas as pd
import numpy as np
import scipy
import seaborn as sns
import scipy.io as sio
import scanpy.external as sce
import matplotlib.pyplot as plt
```

## data preprocessing

```python
adata=sc.read_h5ad("/scratch200/reutj/notebooks/updated_cortex_data_hg38.h5ad")
```

```python
adata
```

```python
sc.pp.filter_cells(adata, min_genes=200)
sc.pp.filter_genes(adata, min_cells=3)
```

```python
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

### pca
harmony uses pca to integrate data so for that fisrt we caculate the pcs..

```python
sc.pp.highly_variable_genes(adata)
# This saves the original set of genes 
adata.raw = adata

adata = adata[:,adata.var.highly_variable]
sc.pp.scale(adata, max_value=10)
```

```python
sc.tl.pca(adata, svd_solver='arpack')
```

```python
sc.pl.pca_variance_ratio(adata, log=True, n_pcs=50, save='') # scanpy generates the filename automatically
```

```python
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=50)
sc.tl.umap(adata)
sc.tl.leiden(adata, resolution=0.5)
```

## integrate on the batch effect - Chemistry 

```python
sce.pp.harmony_integrate(adata, 'Chemistry')
```

```python
#now we change the pca according to harmony calculation and do  the umap agaian
```

```python
adata.obsm['X_pca'] = adata.obsm['X_pca_harmony']
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=30)
sc.tl.umap(adata)
sc.tl.leiden(adata, resolution=0.5)
```

```python
#corrected
sc.pl.umap(adata, color=['Chemistry'],
show=False, palette=sns.color_palette("husl", 2),
    legend_fontsize=6, frameon=True, title='Leiden')
```
