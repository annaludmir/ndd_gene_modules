we will take only samples from multiome_atac that have RNA-seq data from cortex there are 4 subsamples like this with these names:

```python
cortex_only=['BRC2330','BRC2329','HDBR15853','HDBR15777']
sample_cortex=['10X346_2','10X370_1','10X402_1','10X406_7']
```

```python
#lets view the extracted tissue- here telencelphalon instead of cortex but its fine
adata.obs.loc[(adata.obs.Name.isin(sample_cortex))&(adata.obs.Donor.isin(cortex_only)),'Tissue']
```

```python
adata.var_names_make_unique()
```

```python
adata.obs.reset_index(drop=True, inplace=True)
```

```python
adata[(adata.obs.Name.isin(sample_cortex))&(adata.obs.Donor.isin(cortex_only)),:]
```

```python
condition=(adata.obs.Name.isin(sample_cortex))&(adata.obs.Donor.isin(cortex_only))
```

```python
condition = condition.astype(bool)
```

```python
#in order to fix and be able to subset the data we would convert the layer type of all layers to float32
for layer_name in adata.layers:
    print(layer_name, adata.layers[layer_name].dtype)  # Check dtype
    if adata.layers[layer_name].dtype!=np.float32:
        adata.layers[layer_name] = adata.layers[layer_name].astype(np.float32)
```

```python
cortex_adata=adata[condition,:].copy()
```

```python
cortex_adata
```

```python
#write to file
cortex_adata.write_h5ad("/scratch200/reutj/data/cortex_data_to_add.h5ad")
```
