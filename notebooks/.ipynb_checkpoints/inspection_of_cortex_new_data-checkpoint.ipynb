## inspection of the cortex new data

```python
import scanpy as sc
import pandas as pd
import numpy as np
```

```python
adata1= sc.read_h5ad("/scratch200/reutj/data/cortex_data_to_add.h5ad")
```

```python
adata1
```

```python
adata1.layers["raw"]
```

```python
#in order to perform QC and inspect the data we would need raw unnormalized counts so we will
#extract just the raw data as the main data type in adata.X the original data will be stored in "orig" layer
import anndata as ad
adata = ad.AnnData(adata1.layers["raw"])
adata.obs=adata1.obs
adata.var=adata1.var
adata.obsn=adata1.obsm
adata.varm=adata1.varm
```

```python
adata.layers["orig"]=adata1.X.copy()
for i in adata1.layers:
    if i!="raw":
        adata.layers[i]=adata1.layers[i].copy()
```

```python
adata
```

### QC

```python
# mitochondrial genes, "MT-" for human, "Mt-" for mouse
adata.var["mt"] = adata.var_names.str.startswith("MT-")
# ribosomal genes
adata.var["ribo"] = adata.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
adata.var["hb"] = adata.var_names.str.contains("^HB[^(P)]")
```

```python
sc.pp.calculate_qc_metrics(
    adata, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```

```python
adata.var_names_make_unique()
```

```python
sc.pl.violin(
    adata,
    ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
    jitter=0.4,
    multi_panel=True,
)
#this looks like after qc- good. it also looks like upper cells were removed here
```

* results here seem very low and cut from above , unsure wether the chimestry here is v3
cause then we would see much more than 500 counts in some cells
we can also see that the data has already gone some filtering: mt counts are zero

- lets continue with qc - filterd lowly expressed data

```python
sc.pp.filter_cells(adata, min_genes=100)
sc.pp.filter_genes(adata, min_cells=3)
```

```python
adata
```
