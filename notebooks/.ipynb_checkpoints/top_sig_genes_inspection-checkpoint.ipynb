### gsea trial

```python
import gseapy as gp
```

```python
#first plot the most sig results
```

```python
npcs_ges=npcs_ges.sort_values(by="ges_score",ascending=False)
npcs_ges_en=npcs_ges.ges_score
npcs_ges_en.index=npcs_ges.gene
npcs_ges_en
```

```python
gsea = gp.prerank(rnk=npcs_ges_en,
               gene_sets="/scratch200/reutj/data/hsg_gene_lists/custom_hsg_list.gmt",  # Change to your gene set file
               outdir='gsea_results',  # Output directory
               min_size=15,
               max_size=500,
               permutation_type='geneset',
               format='pdf')
```

```python
gsea.res2d
```

```python
terms = gsea.res2d.Term
axs = gsea.plot(terms=terms[0])
```

```python
neuron_ges=neuron_ges.sort_values(by="ges_score",ascending=False)
neuron_ges_en=neuron_ges.ges_score
neuron_ges_en.index=neuron_ges.gene
neuron_ges_en
```

```python
gsea = gp.prerank(rnk=neuron_ges_en,
               gene_sets="/scratch200/reutj/data/hsg_gene_lists/custom_har_genes_list.gmt",  # Change to your gene set file
               outdir='gsea_results',  # Output directory
               min_size=15,
               max_size=500,
               permutation_type='geneset',
               format='pdf')
```

#### trying gsea on deseq2 data

```python
#deseq2 maybe the ranking will be better
deseq_neuron=pd.read_csv("/scratch200/reutj/data/deseq2/deseq2_neuron.csv")
```

```python
#ranking based on padj and log2fc
deseq_neuron['ranking']=-np.log10(deseq_neuron.padj)*deseq_neuron.log2FoldChange
deseq_neuron
```

```python
deseq_neuron_sig=deseq_neuron[(deseq_neuron.padj<0.05)&(deseq_neuron.log2FoldChange>1.5)&(deseq_neuron.baseMean>5)]
```

```python
#all the list of sig
deseq_neuron_en=deseq_neuron_sig.ranking
deseq_neuron_en.index=deseq_neuron_sig.Gene
res=gp.prerank(rnk=deseq_neuron_en,gene_sets=['KEGG_2021_Human','MSigDB_Hallmark_2020','GO_Biological_Process_2021'],seed=2,min_size=10,max_size=700)
```

```python
res.res2d
```

```python
#har sig
gene_list4_en=deseq_neuron_sig.loc[list(deseq_neuron_sig.Gene.isin(har_genes["HAR-associated genes.1"].tolist())),'ranking']
gene_list4_en.index=deseq_neuron_sig.loc[list(deseq_neuron_sig.Gene.isin(har_genes["HAR-associated genes.1"].tolist())),'Gene']
res=gp.prerank(rnk=gene_list4_en,gene_sets=['KEGG_2021_Human','MSigDB_Hallmark_2020','GO_Biological_Process_2021'],seed=5,min_size=5,max_size=700)
```
