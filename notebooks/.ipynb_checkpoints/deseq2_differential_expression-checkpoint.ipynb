### first analysis criteria - cell type 

* we start off by seeing the groups of cells that we have in pur datset based on cell type (conditopn) and donor (batch) the idea is to coninue only with groups  that contain more than 30 cells 

```python
adata_v3.obs.groupby(['Donor','CellClass']).size() 
#XDD:348 basicly has no cells so we will skip him
#also glioblasts in BRC2061,BRC2191,XDD:400 and neurons in XDD:400 and BRC2061 
```

### create the pseudobulks
here the purpose is to build pseudobulk adata in which each group of cells is clustered by getting sum of expression in each gene across all the cewlls in the group to create bulk-like sample that represent relevant chunk of data.
each group is counted as one anndata and then connected together bu anndata.concat

```python
pbs=[]
for sample in adata_v3.obs.Donor.unique():
    if sample!='XDD:348':
        adata_subset=adata[adata.obs.Donor==sample]
    for ct in adata_subset.obs.CellClass.unique():
        final_subset=adata_subset[adata_subset.obs.CellClass==ct]
        if final_subset.shape[0]>30:
            rep_adata=sc.AnnData(X=final_subset.X.sum(axis=0), var=final_subset.var[[]])
            rep_adata.obs_names=[sample+"_"+ct]
            rep_adata.obs['sample']=[sample]
            rep_adata.obs['Cell_type']=[ct]
            pbs.append(rep_adata)
```

```python
ps=sc.concat(pbs)
```

```python
ps
```

```python
#obs table here is the design table for the matrix
ps.obs
```

```python
#add individual cell categories + proliferating/differntiating criteria
ps.obs['RG_or_not']=np.where(ps.obs['Cell_type']=='Radial glia','RG','other')
ps.obs['IPC_or_not']=np.where(ps.obs['Cell_type']=='Neuronal IPC','IPC','other')
ps.obs['Neuroblast_or_not']=np.where(ps.obs['Cell_type']=='Neuroblast','Neuroblast','other')
ps.obs['Glioblast_or_not']=np.where(ps.obs['Cell_type']=='Glioblast','Glioblast','other')
ps.obs['Neuron_or_not']=np.where(ps.obs['Cell_type']=='Neuron','Neuron','other')
ps.obs['proliferating']=np.where(ps.obs['Cell_type'].isin(['Radial glia','Glioblast','Neuronal IPC']),'proliferating','differentiting')
```

```python
#fisher_exact test:
#function
from scipy.stats import hypergeom
from scipy.stats import fisher_exact

def fisher_test_hsg(hsg_genes,sig_table,n_genes=30251,n_hsg=443):
    hsg_intersection=list(set(sig_table.index) & set(hsg_genes))
    table=np.array([[len(hsg_intersection),n_hsg-len(hsg_intersection)],
                    [len(sig_table)-len(hsg_intersection),n_genes-len(sig_table)-len(hsg_intersection)]])
    res = fisher_exact(table, alternative='two-sided')
    return (hsg_intersection , res.pvalue)
```

#### create cell cycle pseudobulk
*here i would like to do the same as in cellclass but with cellcycle

```python
adata_v3.obs.groupby(['Donor','CellCyclePhase']).size()
```

```python
#also here remove xdd:348 and keep the others
```

```python
pbs=[]
for sample in adata_v3.obs.Donor.unique():
    if sample!='XDD:348':
        adata_subset=adata[adata.obs.Donor==sample]
    for cy in adata_subset.obs.CellCyclePhase.unique():
        final_subset=adata_subset[adata_subset.obs.CellCyclePhase==cy]
        if final_subset.shape[0]>30:
            rep_adata=sc.AnnData(X=final_subset.X.sum(axis=0), var=final_subset.var[[]])
            rep_adata.obs_names=[sample+"_"+cy]
            rep_adata.obs['sample']=[sample]
            rep_adata.obs['Cell_cycle_phase']=[cy]
            pbs.append(rep_adata)
```

```python
#add individual cell categories + proliferating/differntiating criteria
ps.obs['Non-cycling_or_not']=np.where(ps.obs['Cell_cycle_phase']=='Non-cycling','Non-cycling','other')
ps.obs['G1_or_not']=np.where(ps.obs['Cell_cycle_phase']=='G1','G1','other')
ps.obs['G2M_or_not']=np.where(ps.obs['Cell_cycle_phase']=='G2M','G2M','other')
ps.obs['S_or_not']=np.where(ps.obs['Cell_cycle_phase']=='S','S','other')
ps.obs['PostM_or_not']=np.where(ps.obs['Cell_cycle_phase']=='PostM','PostM','other')
ps.obs['SG2M_or_not']=np.where(ps.obs['Cell_cycle_phase'].isin(['S','G2M']),'SG2M','other')
```
