## significant enrichments

in order to achieve gene enrichment we will look for gene enrichments in the cortex data
to understand genes that are changed between cell states or types

specificity:

sig(in specific condition)=(sum expression over the condition /number of cells in the condition) / (sum expression other /number of cells not in the condition)

```python
def get_per_exp_o(adata,counts_array,obs_col,condition):
    count=0
    per_exp_o=[]
    n_condition=sum(adata.obs[obs_col]==condition)
    n_other=len(adata.obs)-n_condition
    for gene in gene_sym_adata:
        num_cells_exp=np.count_nonzero(counts_array[:,adata.var_names==gene])
        count+=1
        if count%1000==0:
            print(count) # to keep track of progress
        if num_cells_exp >(n_condition/10):
            gene_exp_o=counts_array[adata.obs[obs_col]!=condition,adata.var_names==gene]
            per_exp_o.append((np.count_nonzero(gene_exp_o))/n_other)
    return np.mean(per_exp_o)
```

```python
#this is a function to get specificty score for any condition given the array matrix and the condition specified
def get_specificity_adjusted(adata,counts_array,obs_col,condition,k,c,alpha,epsilon=1e-06):
    ##this function takes the data with count matrix and gets speceficty scores + other important measures
    ## depending on certain condition. spec score is sombined from exp_ration_mean changes betwen the
    ## condition and other cells , ratio chnages of the percent of cells expressed from the condition or other cells.
    ## we used here sigmoid function tocontrol the effect of percent_other in low valies and **2 to percent condition
    ## to give it more weight, also gene expression in the condition cells is weighthed here using the log function
    specificity={"gene":[],"mean_exp":[],"exp_ratio":[],"per_exp_con":[],"per_exp_other":[],"spec_score":[]}
    n_condition=sum(adata.obs[obs_col]==condition)
    n_other=len(adata.obs)-n_condition
    count=0
    for gene in gene_sym_adata:
        num_cells_exp=np.count_nonzero(counts_array[:,adata.var_names==gene])
        count+=1
        if count%1000==0:
            print(count) # to keep track of progress
        if num_cells_exp >(n_condition/10): #maybe avoid that and consider all?
            #limit to cellswith some expression - more than 1/10 cells based on the population size expressing
            #to avoid unnecessary genes
            gene_exp_con=counts_array[adata.obs[obs_col]==condition,adata.var_names==gene]
            gene_exp_o=counts_array[adata.obs[obs_col]!=condition,adata.var_names==gene]
            mean_exp=np.mean(counts_array[:,adata.var_names==gene])
            mean_exp_con=(np.sum(gene_exp_con))/n_condition
            mean_exp_o=(np.sum(gene_exp_o))/n_other
            per_exp_con=(np.count_nonzero(gene_exp_con))/n_condition
            per_exp_o=(np.count_nonzero(gene_exp_o))/n_other
            adjusted_percent_other = 1 / (1 + np.exp(-k * (per_exp_o - c)))
            specificity_score= (mean_exp_con/(mean_exp_o+epsilon))*((per_exp_con**alpha)/(adjusted_percent_other+epsilon))*np.log1p(mean_exp_con)
            specificity["gene"].append(gene)
            specificity["mean_exp"].append(mean_exp)
            specificity["exp_ratio"].append(mean_exp_con/(mean_exp_o+epsilon))
            specificity["per_exp_con"].append(per_exp_con)
            specificity["per_exp_other"].append(per_exp_o)
            specificity["spec_score"].append(specificity_score)
    return specificity
```

```python
def get_specificity_only(adata,counts_array,obs_col,condition):
    specificity={"gene":[],"mean_exp":[],"exp_ratio":[],"per_exp_con":[],"per_exp_other":[],"spec_score":[]}
    n_condition=sum(adata.obs[obs_col]==condition)
    n_other=len(adata.obs)-n_condition
    count=0
    for gene in gene_sym_adata:
        num_cells_exp=np.count_nonzero(counts_array[:,adata.var_names==gene])
        count+=1
        if count%1000==0:
            print(count) # to keep track of progress
        if num_cells_exp >(n_condition/10): #maybe avoid that and consider all?
            #limit to cellswith some expression - more than 1/10 cells based on the population size expressing
            #to avoid unnecessary genes
            gene_exp_con=counts_array[adata.obs[obs_col]==condition,adata.var_names==gene]
            gene_exp_o=counts_array[adata.obs[obs_col]!=condition,adata.var_names==gene]
            mean_exp=np.mean(counts_array[:,adata.var_names==gene])
            mean_exp_con=(np.sum(gene_exp_con))/n_condition
            mean_exp_o=(np.sum(gene_exp_o))/n_other
            per_exp_con=(np.count_nonzero(gene_exp_con))/n_condition
            per_exp_o=(np.count_nonzero(gene_exp_o))/n_other
            #adjusted_percent_other = 1 / (1 + np.exp(-k * (per_exp_o - c)))
            specificity_score= (mean_exp_con/(mean_exp_o))
            specificity["gene"].append(gene)
            specificity["mean_exp"].append(mean_exp)
            specificity["exp_ratio"].append(mean_exp_con/(mean_exp_o))
            specificity["per_exp_con"].append(per_exp_con)
            specificity["per_exp_other"].append(per_exp_o)
            specificity["spec_score"].append(specificity_score)
    return specificity
```

```python
def compute_ranked_specificity(adata, condition_col):
    """
    Compute specificity scores based on ranks of gene expression.
    
    Args:
    - adata: AnnData object
    - condition_col: str, column in adata.obs specifying cell conditions.
    
    Returns:
    - pd.DataFrame with specificity scores for each gene per condition.
    """
    # Step 1: Rank gene expression in each cell
    ranked_matrix = pd.DataFrame(
        np.apply_along_axis(
            lambda x: pd.Series(x).rank(method="average"), 1, adata.X.toarray()
        ),
        columns=adata.var.index,
        index=adata.obs.index
    )
    print("step 1 finished")

    # Step 2: Group by condition and compute mean rank per gene
    ranked_matrix['condition'] = adata.obs[condition_col].values
    mean_ranks = (
        ranked_matrix.groupby('condition')
        .mean()
        .drop(columns=['condition'])  # Exclude the grouping column
    )

    print("step 2 finished")

    # Step 3: Compute specificity or differential rank scores
    specificity_scores = mean_ranks.apply(
        lambda row: row / row.sum(), axis=1  # Normalize across conditions
    )


    return specificity_scores

# Example usage:
# specificity_scores = compute_ranked_specificity(adata)
# specificity_scores.head()
```

```python
import pandas as pd
import numpy as np

def compute_ranked_specificity_debug(adata, condition_col, batch_size=10000):
    """
    Compute specificity scores based on ranks of gene expression with debug outputs and batching.
    
    Args:
    - adata: AnnData object
    - condition_col: str, column in adata.obs specifying cell conditions.
    - batch_size: int, number of cells to process at once to reduce memory usage.
    
    Returns:
    - pd.DataFrame with specificity scores for each gene per condition.
    """
    n_cells, n_genes = adata.shape

    # Step 1: Rank gene expression in batches
    print("Ranking gene expression...")
    ranked_data = []
    for i in range(0, n_cells, batch_size):
        print(f"Processing batch {i} to {min(i + batch_size, n_cells)}...")
        batch = adata.X[i:i + batch_size].toarray()  # Convert to dense for batch
        ranks = np.apply_along_axis(
            lambda x: pd.Series(x).rank(method="average"), 1, batch
        )
        ranked_data.append(ranks)

    # Combine ranked batches into a full DataFrame
    ranked_matrix = pd.DataFrame(
        np.vstack(ranked_data),
        columns=adata.var.index,
        index=adata.obs.index
    )
    print("Completed ranking.")

    # Step 2: Add condition labels
    print("Adding condition labels...")
    if condition_col not in adata.obs:
        raise ValueError(f"The condition column '{condition_col}' does not exist in adata.obs.")
    
    # Add the condition column explicitly
    ranked_matrix['condition'] = adata.obs[condition_col].values
    print(f"Condition labels added. Unique conditions: {ranked_matrix['condition'].unique()}")

    # Step 3: Compute mean rank per gene for each condition
    print("Computing mean ranks per condition...")
    mean_ranks = (
        ranked_matrix.groupby('condition')
        .mean()
        .drop(columns=['condition'], errors='ignore')  # Exclude the grouping column if present
    )
    print("Mean ranks computation completed.")

    # Step 4: Normalize to compute specificity scores
    print("Computing specificity scores...")
    specificity_scores = mean_ranks.apply(
        lambda row: row / row.sum(), axis=1  # Normalize across conditions
    )
    print("Specificity scores computed.")

    return specificity_scores

# Example usage:
# specificity_scores = compute_ranked_specificity_debug(adata)
# specificity_scores.head()
```
