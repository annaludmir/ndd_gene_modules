```python
$filter for duplicates not found in the original data
dup_gene_names_filt=[i for i in dup_gene_names if i not in hsg1["Gene Name"].tolist()]
```

```python
#get ensembel of these genes
import requests

def get_ensembl_gene_id(gene_symbol):
    """
    Retrieve the Ensembl gene ID based on the gene symbol.
    """
    server = "https://rest.ensembl.org"
    endpoint = f"/xrefs/symbol/homo_sapiens/{gene_symbol}?"
    
    response = requests.get(server + endpoint, headers={"Content-Type": "application/json"})
    if not response.ok:
        response.raise_for_status()
    
    data = response.json()
    # Return the first Ensembl ID found
    return data[0]["id"] if data else None
```

```python
genes_to_add2={}
for gene in dup_gene_names_filt:
    print(gene)
    genes_to_add2[gene]=get_ensembl_gene_id(gene)
```
