## plot results and view in data

```python
import scanpy as sc
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math
```

#### speceficity results enrichments

```python
hsg_gse_ct={'radial glia':0.3128,'IPC':0.1524,'Glioblast':0.7071,'Neuron':0.1091,'Nueroblast':0.2527,'NPCs':0.0196,'proliferating':0.0874,'differentiating':0.2644}
har_gse_ct={'radial glia':0.2308,'IPC':0.264,'Glioblast':0.2469,'Neuron':0.0037,'Nueorblast':0.0172 ,'NPCs':0.2357,'proliferating':0.0451,'differentiating':0.1137}
```

```python
plt.bar(hsg_gse_ct.keys(), [-math.log10(i) for i in list(hsg_gse_ct.values())])
plt.xticks(rotation=45)
plt.ylabel('-log10(padj)')
plt.axhline(y = -math.log10(0.05), color = 'r', linestyle = '-') 
plt.title('GSEA results of human specific genes in cell type spepcificity lists')
```

```python
plt.bar(har_gse_ct.keys(), [-math.log10(i) for i in list(har_gse_ct.values())])
plt.xticks(rotation=45)
plt.ylabel('-log10(padj)')
plt.axhline(y = -math.log10(0.05), color = 'r', linestyle = '-') 
plt.title('GSEA results of HAR genes in cell type spepcificity lists')
```
