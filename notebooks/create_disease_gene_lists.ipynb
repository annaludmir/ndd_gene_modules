```python
#id genes
ID=pd.read_csv("/scratch200/reutj/data/Intellectual_disability.csv")
ID
```

```python
res=pd.DataFrame(mg.querymany(ID["Gene Symbol"], scopes='symbol', fields='ensembl.gene', species='human'))
res
```

```python
res_short=res.iloc[:2729,:]#get onlythe queries that contain genes
```

```python
#genes with no hit
for i in res_short.loc[res_short.ensembl.isna(),'query'].tolist():
    print(i)
```

```python
#i run them to gprofiler (online) to get these results:
results_g=pd.read_csv("/scratch200/reutj/data/gProfiler_hsapiens_10.3.2025_13-07-45.csv")
```

```python
results_g
```

```python
#function to gather the hits for each gene together 
dup_dict={}
for i in range(len(res_short)):
    gene=res['query'][i]
    dup_dict[gene]=[]
for i in range(len(res_short)):
    gene=res['query'][i]
    ens=res.ensembl[i]
    if type(ens)==dict:
        dup_dict[gene].append(ens['gene'])
    elif type(ens)==list:
        dup_dict[gene]=ens

dup_dict
```

```python
no_hit=[]
gprof_genes=results_g.initial_alias.tolist()
for i in gprof_genes:
    ens=results_g.loc[results_g.initial_alias==i,'converted_alias']
    if len(ens)==1:
        if not ens.isna().item():
            print(i,ens.item())
            dup_dict[i]=[ens.item()]
        else:
            print("no hit: ",i)
            no_hit.append(i)
    elif len(ens)>1:
        print("more than one hit: ",i)
```

```python
#more than one hit genes in this list
dup_dict['H3F3A']='ENSG00000163041'
dup_dict['HIST1H4J']='ENSG00000296368'
dup_dict['TCP10L2']='ENSG00000166984'
```

```python
#duplist more than one hit
double_hit=[]
for k,v in dup_dict.items():
    if len(v)>1:
        print(k,":",v)
        double_hit.append(k)
#0,3,0,1,1,0,1,1,1,0,0
```

```python
import requests
# Query Ensembl BioMart
url = "https://rest.ensembl.org/lookup/symbol/homo_sapiens/{}?expand=0"

results = []
for gene in double_hit:
    print(gene)
    response = requests.get(url.format(gene), headers={"Content-Type": "application/json"})
    if response.ok:
        data = response.json()
        results.append({"Gene Symbol": gene, "ENSG ID": data["id"]})
    else:
        results.append({"Gene Symbol": gene, "ENSG ID": "Not Found"})

# Convert to DataFrame
df = pd.DataFrame(results)
print(df)
```

```python
for i in range(len(df)):
    dup_dict[df["Gene Symbol"][i]]=df["ENSG ID"][i]
```

## autism genes

```python
#autism genes

autism=pd.read_csv("/scratch200/reutj/data/SFARI-Gene_genes_01-13-2025release_02-17-2025export.csv")
```

```python
autism
```

```python
#we notice we have gene score nan - those stand for syndromic diseases with augism so will replace nana with s
#and convert to caterogical column
```

```python
autism["gene-score"]=autism["gene-score"].astype(str)
```

```python
autism.loc[autism["gene-score"]=="nan","gene-score"]="S"
autism.loc[autism["gene-score"]=="1.0","gene-score"]="1"
autism.loc[autism["gene-score"]=="2.0","gene-score"]="2"
autism.loc[autism["gene-score"]=="3.0","gene-score"]="3"
```

```python
autism["gene-score"]=autism["gene-score"].astype('category')
```

```python
#for the updated list, just keep ens, gene and gene-score and rename accordingly
```

```python
autism_update=autism.loc[:,['ensembl-id','gene-symbol','gene-score']]
```

```python
autism_update=autism_update.rename(columns={'ensembl-id':'ens','gene-symbol':'gene'})
```

```python
autism_update.to_csv("/scratch200/reutj/data/hsg_gene_lists/autism_genes.csv")
```

```python
#to gmt
output_file="/scratch200/reutj/data/hsg_gene_lists/autism_genes.gmt"
with open(output_file, 'w') as f:
    # Create a line in GMT format for one gene set
    line = f"autism_genes\tDescription\t{'\t'.join(autism_update.gene.tolist())}\n"
    f.write(line)
```
